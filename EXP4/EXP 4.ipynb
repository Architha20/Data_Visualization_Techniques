```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("healthcare-dataset-stroke-data.csv")
```

```python
df.head()
```

```python
print("Shape of Dataset:", df.shape)
```

```python
print(df.dtypes)
```

```python
df.info()
```

```python
df.describe()
```

```python
age_above_60 = df[df["age"] > 60]
age_above_60
```

```python
stroke_patients = df[df["stroke"] == 1]
stroke_patients
```

```python
sorted_bmi = df.sort_values(by="bmi", ascending=False)
sorted_bmi.head()
```

```python
print("Mean =", df["age"].mean())
print("Median =", df["age"].median())
print("Mode =")
print(df["age"].mode())
print("Standard Deviation =", df["age"].std())
```

```python
df.describe(include="all")
```

```python
gender_avg = df.groupby("gender")["age"].mean()
print(gender_avg)
```

```python
work_avg = df.groupby("work_type")["bmi"].mean()
print(work_avg)
```

```python
corr = df[["age", "avg_glucose_level", "bmi", "stroke"]].corr()
print(corr)
```

```python
gender_avg.plot(kind="bar")

plt.title("Average Age by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Age")
plt.show()
```

```python
work_avg.plot(kind="bar")

plt.title("Average BMI by Work Type")
plt.xlabel("Work Type")
plt.ylabel("Average BMI")
plt.show()
```

```python
plt.scatter(df["age"], df["avg_glucose_level"])

plt.title("Age vs Average Glucose Level")
plt.xlabel("Age")
plt.ylabel("Average Glucose Level")
plt.show()
```

```python
plt.hist(df["age"], bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
